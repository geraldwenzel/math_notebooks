# Modeling with Equations
Numerous problems from engineering, science, finance, and other fields can be modeled using equations. In many cases, algebra alone can be used solve these problems. 

### Technical Discussion Questions

- **What are the advantages of using algebraic equations to model real-world problems?**  
The advantages of using algebraic equations to model real-world problems include the ability to systematically analyze relationships between variables, predict outcomes, and solve for unknowns in a structured way. Algebraic models can also be manipulated to explore different scenarios and provide insights into the behavior of systems.

- **How can visualizing a problem through diagrams or graphs aid in understanding the relationships between variables?**  
Visualizing a problem through diagrams or graphs can aid in understanding the relationships between variables by providing a clear representation of how different quantities interact. Diagrams can illustrate spatial relationships, while graphs can show trends and patterns, making it easier to identify key features of the problem and potential solutions.  

- **What role does checking the solution play in the problem-solving process?**  
Checking the solution plays a crucial role in the problem-solving process as it helps to verify the accuracy of the answer obtained. By substituting the solution back into the original equation, one can confirm that the solution satisfies the conditions of the problem. This step ensures that any errors made during the modeling or solving process are identified and corrected, leading to greater confidence in the results.

- **What are some common pitfalls to avoid when translating verbal descriptions into algebraic equations?**  
  - Not understanding the relationship between the known and unknown quantities will result in an algebraic equation that does not correctly model the problem.  
  - Misinterpreting the problem statement can lead to incorrect assumptions about the variables involved.  
  - Failing to define all relevant variables can result in an incomplete model.  
  - Neglecting to check the solution can lead to accepting incorrect answers.  

- **Why is identification of the fact given in the problem that defines the relationship between the known and unknown quantities important?**  
Understanding the relationship between the known and unknown quantities is critical to correctly model the problem as an algebraic equation.

### Height of 2 Adjacent and Congruent Right Triangles
Two right triangles are adjacent to each other. Both triangles are congruent. The triangles share a common side that makes up the height of both triangles. The hypothenuse of each triangle is 5 ft longer than the height of the triangles. The base of both triangles combined is equal to the length of the hypothenuse. Find the height of the triangles to the nearest inch.

Let $x$ be the height of the triangles.

- The hypotenuse of each triangle is $x + 5$ ft. 
- The base of both triangles is $x + 5$ ft.
- The base of a single triangle is $\frac{x+5}{2}$ ft.

```python
fig, ax = plt.subplots()

x_vals = [0, 8]
y_vals = [6, 0]
ax.plot(x_vals, y_vals, label='Hypotenuse', color='blue', linestyle='--')

x_vals = [0, -8]
y_vals = [6, 0]
ax.plot(x_vals, y_vals, label='Hypotenuse', color='blue', linestyle='--')


x_vals = [-8, 8]
y_vals = [0, 0]
ax.plot(x_vals, y_vals, label='Base', color='red', linestyle='--', linewidth=3)

# draw a vertical line from the top of the hypotenuse to the base
x_vals = [0, 0]
y_vals = [6, 0]
ax.plot(x_vals, y_vals, label='Height', color='green', linestyle='--', linewidth=3)

y_max = 8
ax.axis(xmin=-10, xmax=10, ymin=0, ymax=y_max)
ax.grid()
ax.set_xlabel('')
ax.set_ylabel('')

# remove x-axis ticks
ticks = ax.get_xticks()
labels = ['' for t in ticks]
ax.set_xticks(ticks)
ax.set_xticklabels(labels)

# remove y-axis ticks
ticks = ax.get_yticks()
labels = ['' for t in ticks]
ax.set_yticks(ticks)
ax.set_yticklabels(labels)

# remove the grid lines
ax.grid(False)

# remove the ticks
ax.tick_params(left=False, bottom=False)

ax.legend(loc='upper left', fontsize=10, frameon=False);

ax.annotate(
    'x + 5',
    xy=(4, 3),         # point on the hypotenuse to annotate
    xytext=(5, 5),     # position of the text
    arrowprops=dict(arrowstyle='->', color='black'),
    fontsize=12
)

ax.annotate(
    'x + 5',
    xy=(-4, 3),         # point on the hypotenuse
    xytext=(-7, 5),     # position of the text
    arrowprops=dict(arrowstyle='->', color='black'),
    fontsize=12
)

ax.annotate(
    'x + 5',
    xy=(3, 0),         # a point on the base
    xytext=(3, 1.5),   # position of the text above the base
    arrowprops=dict(arrowstyle='->', color='black'),
    fontsize=12,
    ha='center'
)

ax.annotate(
    'x',
    xy=(0, 3),         # point on the Y-axis
    xytext=(-1.5, 4),   # position of the label
    arrowprops=dict(arrowstyle='->', color='black'),
    fontsize=12
);
```

Because the two adjacent triangles are right triangles, the pythagorean theorem can be used to express the length of the base and hypotenuse in terms of $x$.

$x^2 + \left(\frac{x+5}{2}\right)^2 = (x + 5)^2$

$$
\begin{aligned}
x^2 + \left(\frac{x+5}{2}\right)^2 &= (x + 5)^2 \\
x^2 + \frac{(x+5)^2}{4} &= (x + 5)^2 \\
4x^2 + (x+5)^2 &= 4(x + 5)^2 \\
4x^2 + (x^2 + 10x + 25) &= 4(x^2 + 10x + 25) \\
4x^2 + x^2 + 10x + 25 &= 4x^2 + 40x + 100 \\
5x^2 + 10x + 25 &= 4x^2 + 40x + 100 \\
5x^2 + 10x + 25 - 4x^2 - 40x - 100 &= 0 \\
x^2 - 30x - 75 &= 0 \\
\end{aligned}
$$

Solve for $x$ using the quadratic formula:
$$
\begin{aligned}
x &= \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} \\
x &= \frac{-(-30) \pm \sqrt{(-30)^2 - 4(1)(-75)}}{2(1)} \\
x &= \frac{30 \pm \sqrt{900 + 300}}{2} \\
x &= \frac{30 \pm \sqrt{1200}}{2} \\
x &= \frac{30 \pm 20\sqrt{3}}{2} \\
x &= 15 \pm 10\sqrt{3} \\
\end{aligned}
$$

Approximate the value of $x$:
$$
\begin{aligned}
x &\approx 15 + 17.32 \approx 32.32 \text{ ft} \\
x &\approx 15 - 17.32 \approx -2.32 \text{ ft (not valid)}
\end{aligned}
$$

Because the height cannot be negative, only the positive solution is valid.  
Therefore, the height of the triangles is approximately $32.32$ ft.
Converting to the nearest inch:
$$
32.32 \text{ ft} \cdot \frac{12 \text{ in}}{1 \text{ ft}} \approx 387.84 \text{ in} \approx 388 \text{ in}
$$

$\boxed{\text{The height of the triangles is approximately 388 inches.}}$

### Shadow Length
A streetlight is 6 m tall. 10 m away from the streetlight, a 2 m tall person is standing. How long is the shadow of the person?

Let $x$ be the length of the person's shadow.

The length of the person's shadow is the only unknown.

```python
fig, ax = plt.subplots()

x_vals = [0, 15]
y_vals = [6, 0]
ax.plot(x_vals, y_vals, label='Light from streetlight', color='blue', linestyle='--')

x_vals = [0, 10]
y_vals = [2, 2]
ax.plot(x_vals, y_vals, label='Base of smaller triangle', color='red', linestyle='--')

x_vals = [0, 15]
y_vals = [0, 0]
ax.plot(x_vals, y_vals, label='Base of larger triangle', color='blue', linestyle='--', linewidth=3)

x_vals = [10, 10]
y_vals = [0, 2]
ax.plot(x_vals, y_vals, label='Person', color='black')

y_max = 8
ax.axis(xmin=0, xmax=16, ymin=0, ymax=y_max)
ax.grid()
ax.set_xlabel('Distance (m)')
ax.set_ylabel('Height (m)')
ax.axvline(0, color='black', lw=2)
ticks = ax.get_xticks()
labels = [str(int(t)) if t == 10 else '' for t in ticks]
ax.set_xticks(ticks)
ax.set_xticklabels(labels)
ax.legend(loc='upper left', fontsize=10, frameon=False);
```

The top of the streetlight and the top of the person's head form a right triangle. Because the vertical distance from the top of the streetlight to to the person's head is 4 m and the horizontal distance from the streetlight to the person is 10 m, similar triangles can be used to find the length of the person's shadow.

$$
\begin{aligned}
\frac{4}{10} &= \frac{6}{10 + x} \\
4(10 + x) &= 6(10) \\
4(10) + 4x &= 60 \\
40 + 4x &= 60 \\
4x &= 60 - 40 \\
4x &= 20 \\
x &= \frac{20}{4} \\
x &= 5
\end{aligned}
$$

```python
eq = Eq(Rational(4, 10), 6/(10+x))
display(eq)
```

```python
solveset(eq, x, domain=S.Reals)
```

$\boxed{\text{The length of the person's shadow is 5 m.}}$

### Height of a Wall
A person with an eye level 5 ft above the ground is standing 25 ft from a 20 ft tall wall. 125 ft beyond the wall is another wall. How tall is the second wall if the person can see the top of the first wall and the top of the second wall in a straight line?

Let $h$ be the height of the second wall.

The height of the first wall is the only unknown.

```python
fig, ax = plt.subplots()

x_vals = [0, 150]
y_vals = [5, 95]
ax.plot(x_vals, y_vals, label='Line of sight', color='blue', linestyle='--')

x_vals = [0, 150]
y_vals = [5, 5]
ax.plot(x_vals, y_vals, label='Eye level', color='red', linestyle='--')

x_vals = [25, 25]
y_vals = [0, 20]
ax.plot(x_vals, y_vals, label='First wall', color='black')

x_vals = [150, 150]
y_vals = [0, 95]
ax.plot(x_vals, y_vals, label='Second wall', color='brown')

y_max = 110
ax.axis(xmin=0, xmax=160, ymin=0, ymax=y_max)
ax.grid()
ax.set_xlabel('Distance (ft)')
ax.set_ylabel('Height (ft)')
ax.axhline(0, color='black', lw=2)
ax.axvline(0, color='black', lw=2)
ax.legend(loc='upper left', fontsize=10, frameon=False);
```

Similar triangles are formed by the two walls and the line of sight. Similar triangles allow us to set up a proportion between the heights and distances of the two walls, allowing us to solve for the height of the second wall.

$$\frac{15}{25} = \frac{h}{150}$$

$$
\begin{aligned}
\frac{15}{25} &= \frac{h}{150} \\
15 \cdot 150 &= 25 \cdot h \\
2250 &= 25h \\
h &= \frac{2250}{25} \\
h &= 90
\end{aligned}
$$

$\boxed{\text{Because the eye level of the person is 5 ft above the ground, the height of the second wall is 90 + 5 = 95 ft.}}$

### Dimensions of a Box
A box has a volume of 180 cubic feet. The length is 9 feet greater than its height, and its width is 4 feet less than its height. What are the dimensions of the box?  

```python
eq = Eq(180, (x+9)*(x)*(x-4))
display(eq)
```

```python
[_ for _ in solve(eq, domain=S.Reals) if _.is_real]
```

$x$ = height of the box
$$
\begin{aligned}
180 &= (x + 9)(x - 4)x \\
180 &= (x^2 + 5x - 36)x \\
180 &= x^3 + 5x^2 - 36x \\
x^3 + 5x^2 - 36x - 180 &= 0 \\
x^2(x + 5) - 36(x + 5) &= 0 \\
(x + 5)(x^2 - 36) &= 0 \\
(x + 5)(x - 6)(x + 6) &= 0 \\
\end{aligned}
$$

Because the height cannot be negative, $x = 6$.  

The dimensions of the box are:
- height: $x = 6$ feet
- length: $x + 9 = 15$ feet
- width: $x - 4 = 2$ feet

### Group Investment
A group of investors decides to buy a painting for $120,000. Each investor is going to contribute the same amount. If one more investor can be found, each investor's contribution will reduce by $6000. How many investors are in the initial group?

**Identify the unknown quantity the problem is asking you to find.**  
How many investors are in the initial group?

**Assign a variable to represent the unknown quantity. In many cases, this will be $x$.**  
Let $x$ be the number of investors in the initial group.

**Identify other unknowns by reading through the problem again. Express all other unknown quantities in terms of the variable that represents the unknown quantity the problem is asking you to find.**  
The number of investors in the initial group is the only unknown.

**Identify the fact given in the problem that defines the relationship between the known and unknown quantities.**  
The addition of one more investor reduces each investor's contribution by $6000.

**Set up the model using algebra by writing an equation that expresses this relationship.**  
$$\frac{120000}{x}-6000 = \frac{120000}{x+1}$$

**Solve the equation for the variable that represents the unknown quantity the problem is asking you to find.**  
$$
\begin{aligned}
\frac{120000}{x}-6000 &= \frac{120000}{x+1} \\[1em]
\frac{120}{x}-6 &= \frac{120}{x+1} \\[1em]
x(x+1)\left(\frac{120}{x}-6\right) &= x(x+1)\left(\frac{120}{x+1}\right) \\[1em]
120(x+1)-6x(x+1) &= 120x \\[1em]
120x+120-6x^2-6x &= 120x \\[1em]
6x^2+6x-120 &= 0 \\[1em]
x^2+x-20 &= 0 \\[1em]
(x+5)(x-4) &= 0 \\[1em]
x &= \{-5, 4\}
\end{aligned}
$$

**State the conclusion by expressing the solution as a sentence that answers the question posed in the problem.**  
Because the number of investors cannot be negative, the initial group consists of 4 investors.

### Radius of a Sphere
**If three solid spheres of silver, of radius 2 mm, 3 mm and 4 mm are melted down and recast into a single solid sphere. What is the radius of the new sphere?**

The volume of a sphere is given by the formula:

$V = \frac{4}{3} \pi r^3$

where $r$ is the radius of the sphere.

**Compute the volume of each sphere:**

**Sphere with radius 2 mm:**  
$V_2 = \frac{4}{3} \pi (2)^3 = \frac{4}{3} \pi (8) = \frac{32}{3} \pi$

**Sphere with radius 3 mm:**  
$V_3 = \frac{4}{3} \pi (3)^3 = \frac{4}{3} \pi (27) = \frac{108}{3} \pi = 36\pi$

**Sphere with radius 4 mm:**  
$V_4 = \frac{4}{3} \pi (4)^3 = \frac{4}{3} \pi (64) = \frac{256}{3} \pi$

**Compute the Total Volume**
Since the jeweler melts the spheres together, the total volume is:

$V_{\text{total}} = V_2 + V_3 + V_4$

$V_{\text{total}} = \frac{32\pi}{3} + 36\pi + \frac{256\pi}{3}$

Convert $36\pi$ to have a denominator of 3:

$36\pi = \frac{108\pi}{3}$

$V_{\text{total}} = \frac{32\pi}{3} + \frac{108\pi}{3} + \frac{256\pi}{3}$

$V_{\text{total}} = \frac{396\pi}{3} = 132\pi$

**Find the radius of the new sphere:**  
Let $R$ be the radius of the new sphere. The volume of the new sphere is:

$$
\begin{aligned}
\frac{4\pi R^3}{3} &= 132\pi \\
\frac{1}{\pi}\left(\frac{4\pi R^3}{3}\right) &= 132\pi \cdot \frac{1}{\pi} \\
\frac{4R^3}{3} &= 132 \\
3 \cdot \frac{4R^3}{3} &= 132 \cdot 3 \\
4R^3 &= 396 \\ 
\frac{1}{4} \cdot 4R^3 &= 396 \cdot \frac{1}{4} \\ 
R^3 &= 99 \\
\sqrt[3]{R^3} &= \sqrt[3]{99} \\
R &\approx 4.63 \text{ mm}
\end{aligned}
$$

$\boxed{\text{The radius of the new sphere is approximately }4.63 \text{ mm}.}$
